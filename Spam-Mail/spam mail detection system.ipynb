```python
# import dependencies
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
```

data collection and preprocessing


```python
mail_data=pd.read_csv('/content/mail_data.csv')
```

```python
mail_data.head()
```

```python
mail_data['Category'].value_counts()
```

spam--> spam mail  


ham --> not spam mail

```python
mail_data.shape
```

```python
mail_data.isnull().sum()
```

```python
x=mail_data['Message']
y=mail_data['Category']
```

```python
print(x)
print(y)
```

```python
x.shape
```

spliting data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=3)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

feature of text data(converting character data into numerical data and removing of unwanted symbols and grammar)

```python
vectorizer=TfidfVectorizer(stop_words='english')  # it directly remove the stop words
```

```python
x_train=vectorizer.fit_transform(x_train)
x_test=vectorizer.transform(x_test)
```

apply leval encoding for categorical data

```python
Level_encoder=LabelEncoder()
y_train=Level_encoder.fit_transform(y_train)
y_test=Level_encoder.transform(y_test)
```

training model

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

accuracy check

```python
# accuracy check on training data
x_train_prediction=model.predict(x_train)
accuracy=accuracy_score(x_train_prediction,y_train)
print(accuracy)
```

```python
# accuracy check on testing data
x_test_prediction=model.predict(x_test)
accuracy=accuracy_score(y_test,x_test_prediction)
print(accuracy)
```

making predictive system

```python
input=["SMS. ac Sptv: The New Jersey Devils and the Detroit Red Wings play Ice Hockey. Correct or Incorrect? End? Reply END SPTV"]

# convert text to feature vector
input_data_feature=vectorizer.transform(input)
# making prediction
prediction=model.predict(input_data_feature)
print(prediction[0])

if prediction[0]==0:
  print("ham mail")
else:
  print("spam mail")
```
