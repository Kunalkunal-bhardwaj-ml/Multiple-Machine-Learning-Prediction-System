```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
heart_disease=pd.read_csv('/content/HeartDiseaseTrain-Test.csv')
```

apply lebel encoding

```python
level_encoder=LabelEncoder()
```

```python
heart_disease['sex']=level_encoder.fit_transform(heart_disease['sex'])
heart_disease['chest_pain_type']=level_encoder.fit_transform(heart_disease['chest_pain_type'])
heart_disease['fasting_blood_sugar']=level_encoder.fit_transform(heart_disease['fasting_blood_sugar'])
heart_disease['rest_ecg']=level_encoder.fit_transform(heart_disease['rest_ecg'])
heart_disease['exercise_induced_angina']=level_encoder.fit_transform(heart_disease['exercise_induced_angina'])
heart_disease['slope']=level_encoder.fit_transform(heart_disease['slope'])
heart_disease['vessels_colored_by_flourosopy']=level_encoder.fit_transform(heart_disease['vessels_colored_by_flourosopy'])
heart_disease['thalassemia']=level_encoder.fit_transform(heart_disease['thalassemia'])

```

```python
heart_disease.head()
```

apply data standardization

```python
numeric_cols=[ 'age','resting_blood_pressure', 'cholestoral', 'Max_heart_rate','oldpeak']
```

```python
standard_scaler=StandardScaler()
```

```python
heart_disease[numeric_cols]=standard_scaler.fit_transform(heart_disease[numeric_cols])
```

```python
heart_disease.head()
```

```python
heart_disease.tail()
```

split the data into train and test

```python
x=heart_disease.drop(columns='target',axis=1)
y=heart_disease['target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
x_train.shape,x_test.shape,heart_disease.shape
```

train ml model

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

accuracy check

```python
# accuracy check on train data
x_train_prediction=model.predict(x_train)
train_data_accuracy=accuracy_score(x_train_prediction,y_train)
print(train_data_accuracy)
```

```python
# accuracy check on test data
x_test_prediction=model.predict(x_test)
test_data_accuracy=accuracy_score(x_test_prediction,y_test)
print(test_data_accuracy)
```

making a predictive syatem

```python
input_data=(0.503520,	1,	1,	0.479107,	-0.484803,	1,	2,	0.647366,	1,	-0.912329,	0,	4,	0 )
input_data_as_numpy_array=np.asarray(input_data)
input_data_reshaped=input_data_as_numpy_array.reshape(1,-1)
prediction=model.predict(input_data_reshaped)
print(prediction)
if(prediction[0]==0):
  print('patient has not heart_diseasse')
else:
  print('patient has heart_disease')
```

```python

import pandas as pd
import pickle
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# 1. Upload original Heart Disease CSV file
uploaded = files.upload()
file_name = next(iter(uploaded))

heart_disease = pd.read_csv(file_name)

# 2. Categorical columns
categorical_columns = [
    'sex',
    'chest_pain_type',
    'fasting_blood_sugar',
    'rest_ecg',
    'exercise_induced_angina',
    'slope',
    'vessels_colored_by_flourosopy',
    'thalassemia'
]

# 3. Create a separate encoder for each column
heart_encoders = {}

for column in categorical_columns:
    encoder = LabelEncoder()
    encoder.fit(heart_disease[column].dropna())
    heart_encoders[column] = encoder

# 4. Save encoders
with open('heart_disease_encoders.sav', 'wb') as file:
    pickle.dump(heart_encoders, file)

# 5. Download the file
files.download('heart_disease_encoders.sav')
```
